```python
import torch
from d2l import torch as d2l
import math
import os
from ptbData.ptb_loader import *
```

```python
sentences = load_ptb_data() 
vocab = d2l.Vocab(sentences, min_freq=10) 
subsampled, _ = down_sample(sentences, vocab) 
corpus = [vocab[line] for line in subsampled]
corpus = corpus[1:] 
```

```python
import collections

counter = collections.Counter([token for line in corpus for token in line])

idx_freq=list(counter.items())
tokens_freq={" ".join(vocab.idx_to_token[idx]+'_'):freq for idx,freq in idx_freq}

```

```python
symbols_raw = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm',
'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z',
'_', '[UNK]']

symbols={chr:0 for chr in symbols_raw}
priority=1
```

```python
def get_most_freq_token_pairs(tokens_freq):
    pairs=collections.defaultdict(int)
    for token,freq in tokens_freq.items():
        symbols=token.split()
        for i in range(len(symbols)-1):
            pairs[symbols[i],symbols[i+1]]+=freq
    return max(pairs,key=pairs.get)

def merge_symbols(max_pair,tokens_freq,symbols):
    global priority
    new_symbol=max_pair[0]+max_pair[1]
    symbols[new_symbol]=priority
    priority+=1
    tokens_freq={token.replace(' '.join(max_pair),new_symbol):freq for token,freq in tokens_freq.items()}
    return tokens_freq
```

```python
num_merges = 120
for i in range(num_merges):
    max_freq_pair = get_most_freq_token_pairs(tokens_freq)
    tokens_freq = merge_symbols(max_freq_pair, tokens_freq, symbols)
    print(f'合并# {i+1}:',max_freq_pair)
```

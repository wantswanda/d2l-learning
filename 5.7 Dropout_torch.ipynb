```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
net0 = nn.Sequential(nn.Flatten(),nn.Linear(784,256),nn,ReLU(),
    nn.Dropout(0.5),nn.Linear(256,256),nn.ReLU(),nn.Dropout(0.2),nn.Linear(256,10),nn.Softmax(dim=1))


def init_weights(m):
    if isinstance(m, nn.Linear):
        nn.init.normal_(m.weight, std=0.01)
        nn.init.zeros_(m.bias)

net0.apply(init_weights)

loss = nn.CrossEntropyLoss()

trainer = torch.optim.SGD(net0.parameters(),lr=0.1)

train_iter,test_iter = d2l.load_data_fashion_mnist(batch_size=256)

```

```python
def evaluate_loss(net,data_iter,loss):
    metric = d2l.Accumulator(2)
    for X,y in data_iter:
        y_hat = net(X)
        l = loss(y_hat,y)
        metric.add(l.sum(),y.numel())

    return metric[0]/metric[1]
```

```python
def train(train_iter,test_iter,loss,trainer,net,num_epochs):
    animator = d2l.Animator(xlabel='epochs', ylabel='loss',
                            xlim=[0, num_epochs], legend=['train', 'test'])
    for epoch in range(num_epochs):
        for X,y in train_iter:
            loss_ = loss(net(X),y)
            loss_.backward()
            trainer.step()
        animator.add(epoch+1,(evaluate_loss(net,train_iter,loss),evaluate_loss(net,test_iter,loss)))
```

```python
train(train_iter,test_iter,loss,trainer,net0,num_epochs)
```
